### General Principle: Write Less Code

### What if We Need to Repeat Some Process Many, Many Times?

<img src="img/copy_paste.gif">

* e.g. standard Plots, scoring questionnaires, transform data across different collections, computing stats, etc.
* To prevent this turning into wayyy too many lines of code: use _Functions_!

### Declaring Functions supports modular programming.
* Coding would be opaque & difficult to work with if we could only write sequences in serial from top to bottom (e.g., <a href="https://github.com/JeremyHogeveen/dspn_spring2026_github/blob/main/code/dspn_spring2026_week1_01222026.ipynb">our "imperative" PB&J instructions</a>).
* Functions are blocks of code that only run when explicitly called, which can be reconfigured in any number of ways (e.g., <a href="https://github.com/JeremyHogeveen/dspn_spring2026_github/blob/main/code/dspn_spring2026_week1_01222026.ipynb">our "declarative" PB&J instructions</a>).
    * Each function operates independently.
    * Functions then combined together in a workflow.
    * Allow you to build up a complex, but still (reasonably) well-organized codebase.

<img src="img/work_schematic.png" width="650">

### Coding Functions in Python

#### <mark>__def__</mark>

* Use __def__ and __:__ to create a user-defined function
    * Same indenting rules as iterations and conditionals
* __def__ doesn't actually _execute_ the function, it just creates it
    * Instead, subsequent to defining the function we execute it placing any associated inputs in parentheses -- __()__
    * Logic behind our code no longer needs to proceed in serial order.

### Example 1: Let's code a function that adds 5 to an input value

```python
# Defining the function
def add_five(user_input):
    # do some operation
    print(user_input+5)
```

```python
add_five(4)
```

#### Now, say we want to hold onto the output for later

```python
out_value = add_five(4)
print(out_value)
```

#### Why does it say "None"? How can we modify it to store the output?

```python
# Defining the function
def return_add_five(user_input):
    # do some operation
    x = user_input+5
    return x
```

```python
out_value_2 = return_add_five(4)
```

```python
print(out_value) # came from a function with nothing returned
print(out_value_2) # came from a function with x returned
```

#### How could we modify the function to add two user-defined values (i.e., not limited to adding 5)?

```python
# Defining the function
def add_two_integers(in1,in2):
    x = in1 + in2
    return x
```

```python
out_value_3 = add_two_integers(4,5)
```

```python
print(out_value_2) # product of a function that adds five to the input #
print(out_value_3) # product of a functin that added two integers together (in this case, 4 and 5)
```

__Note: We're not limited to evaluating single expressions in a function. We can add control flows and/or nested control flows as well!__

#### Example 2: Write a function to determine if an input value is even or odd

```python
# defining the function
def oddeven(in1):
    remainder = in1 % 2
    if remainder == 0:
        out = "even"
    else:
        out = "odd"
    return out
```

```python
even_odd_test = oddeven(10)
print(even_odd_test)
```

#### Note: Can _return_ different types of objects (e.g. collections, plots, etc.), can also return multiple objects.

```python
# Defining the function
def oddeven_2(in1,in2):
    if in1 % 2 == 0:
        out1 = "even"
    else:
        out1 = "odd"
    if in2 % 2 == 0:
        out2= "even"
    else:
        out2 = "odd"
    x = [out1,out2]
    return x
even_odd_list = oddeven_2(1,2)
print(even_odd_list)
```

```python
# Defining the function
def oddeven_3(in1,in2):
    if in1 % 2 == 0:
        out1 = "even"
    else:
        out1 = "odd"
    if in2 % 2 == 0:
        out2= "even"
    else:
        out2 = "odd"
    return out1, out2
    
even_odd_1,even_odd_2 = oddeven_3(1,2)
print(even_odd_1,even_odd_2)
```
