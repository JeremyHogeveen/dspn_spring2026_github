# Section 1: Working with lists.

* Class _list_.
* General purpose collection of data in Python.
* Stray thoughts:
    * Lists are general purpose, while Numpy arrays are a more fixed / rigid category of list.
    * There are pros and cons / apt use cases for lists and Numpy arrays.
    * Here we'll practice with lists to show you how they work, but I commonly store as Numpy arrays when I'm working with homogeneous data sources.

## List exercises:

* To start, let's use Numpy to make a random number list.
    * E.g. 10 numbers taken from a random normal distribution.

```python
import numpy as np

list_randn = np.random.randn(10)
list_randn = list_randn.tolist()
print(list_randn)
```

* List indexing:
    * Now, pull the 3rd number from the list.

```python
third_num = list_randn[2]
print(third_num)
```

* List slicing:
    * Now, pull the 3rd, 4th, 5th, and 6th numbers from the list.

```python
third_through_sixth = list_randn[2:6]
print(third_through_sixth)
```

* List assignment:
    * Let's make the 5th number in the list equal to 2.30

```python
list_randn[4] = 2.30
print(list_randn)
```

* Appending to a list:
    * Now, let's add three new observations to the list, pulled from a random normal distribution.

```python
# generate new observations
new_obs = np.random.randn(3)
new_obs = new_obs.tolist()
# print(new_obs)

# with append
list_randn_2 = np.append(list_randn,new_obs)
# print(list_randn_2)

# with concatenate
list_randn_3 = np.concatenate([list_randn,new_obs])
# print(list_randn_3)

# with '+'
list_randn_4 = list_randn + new_obs
print(list_randn_4)
```

## 2.1 Dict exercises:

# Section 3: Dot Notation.

* Python uses the dot _<.>_ to indicate that we are accessing data or functionality inside an object.
    * Can only be used used on objects / instances that contain the function that comes after.

## 3.1 Dot notation vs. functional syntax.

* Functional syntax (e.g. _function(input)_) can be used for any functions that are built-in for all objects in Python

```python
phrase = "aPPleS aRe deLiCIous"
len(phrase)
```

* Dot notation can be useful to call on specific methods that are unique to the type of object we're working with.

```python
print(phrase.lower())

# lower(phrase)

# integer = 4
# integer.lower()
```

```python
# dir(phrase)
```

* <mark>Challenge: Tell me one specific method that we could use with floats but not integers?</mark>

```python
# making some variables
floating = 4.15
integer_2 = 4

# find list of methods for each variable type
floating_methods = dir(floating)
integer_methods = dir(integer_2)

# looking for differences between the lists
from collections import Counter
diff_counts = Counter(floating_methods) - Counter(integer_methods)
float_specific_methods = list(diff_counts.elements())
print(float_specific_methods)
```

## 4.3 Iteration using a WHILE loop

* Continues repeating an activity while some condition(s) is(are) true.

```python
# Loop from 0 to 100 in steps of 10

# preamble
x = 0
end = 100
step = 10

# loop
while x <= end:
    print(x)
    x = x + step

print('i am no longer in the loop.')
```

## 4.4 Iteration using a FOR loop

```python
# preamble
x = 0
end = 100
step = 10

# for loop
for i in range(x,end+step,step):
    print(x)
    x += step
```

## 4.5 Iteration using List Comprehension

* Basic syntax: [output for i in list]
* Can also add in conditionals (next section...)

```python
# preamble
x = 0
end = 100
step = 10
[x for x in range(x,end+step,step)]
```
