## 3.4 Subsetting data

### Filter (i.e. subset rows)

<img src="img/filter.png" width="600">

```python
# load pandas as abbreviated "pd"
import pandas as pd

# web source for the data:
website = "https://yeatmanlab.github.io/AFQBrowser-demo/data/subjects.csv"

# create a simple function to skip cols we don't want
def f_skipcol(path_in,list_skip):
    # read all the columns in to get the full list
    list_allcols = pd.read_csv(path_in,nrows=0).columns
    # create a difference list comparing the column names
    list_cols_to_use = list_allcols.difference(list_skip)
    # read in only the cols we want to use
    df_out = pd.read_csv(path_in, usecols = list_cols_to_use)
    return df_out

# using the function to get the data we need
list_pleasedestroy = [ "Unnamed: 0" ]
df_subinfo = f_skipcol(website,list_pleasedestroy)
print(df_subinfo)
```

* Now, subset to participants 20 years and older

```python
print(df_subinfo.Age>=20)
```

## 5. Combining Data Sets

<img src="img/combining_data.png" width="600">

```python
import numpy as np

df1 = pd.DataFrame({'pid' : ['P1','P2','P3'],
                   'dep' : [np.random.randint(25,75),np.random.randint(25,75),np.random.randint(25,75)] })
# print(df1)

df2 = pd.DataFrame({'pid' : ['P1','P2','P4'],
                   'anx' : [np.random.randint(25,75),np.random.randint(25,75),np.random.randint(25,75)] })
# print(df2)

# standard merge #1 -- join matching rows from df2 to df1
print(pd.merge(df1,df2,how='left',on='pid'))

# standard merge #2 -- join matching rows from df1 to df2
print(pd.merge(df1,df2,how='right',on='pid'))

# standard merge #3 -- only keep rows present in BOTH dfs
print(pd.merge(df1,df2,how='inner',on='pid'))

# standard merge #4 -- keep rows present in ANY dfs
print(pd.merge(df1,df2,how='outer',on='pid'))
```
