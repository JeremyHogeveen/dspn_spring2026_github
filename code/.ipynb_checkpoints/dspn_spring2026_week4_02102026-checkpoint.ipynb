## 4.6 iterating through preexisting lists

* Sometimes we may not need to explicitly specify the start, end, and/or step of our loop.
* For example, we may want to iterate through a list of data we already have.

```python
random_stuff = ['eleventy','apple','7.14','banana','88']

for elem in random_stuff:
    print('itering directly thro0ugh the list elements:',elem)

for i in range(0,len(random_stuff)):
    print('printing the ith item in the list:',i,random_stuff[i])
```

* <mark>Challenge: Using the approaches above, determine the length of _random_stuff_ in two ways: 1. using functional syntax (should be 1 line), and 2. Using a for loop (>1 line of code). Bonus: Using list comprehension?.</mark>

```python
# using functional syntax
print(len(random_stuff))
```

```python
# using iteration
count = 0
for elem in random_stuff:
    count += 1
print(count)
```

```python
# using list comprehension
count = 0
count = [(count + 1) for elem in random_stuff if count < len(random_stuff)]
print(len(count))

# my list comprehension for counting
list_comp_counter = [i+1 for i in range(len(random_stuff))]
max(list_comp_counter)
```

* Last iteration exercise: write a loop to update an outcome variable based on a logarithmic scale. Plot it.

```python
#package loading
import numpy as np
import matplotlib.pyplot as plt

# preamble
samples = 50
y = []

# the loop
for i in range(1,samples+1):
    tmp_y = np.log(i)
    y.append(tmp_y)

# plotting the results
plt.plot(y,marker='o',color='b',linestyle='-')
plt.xlabel('Iteration')
plt.ylabel('Outcome Variable')
plt.show()
```

## Conditionals: _if-then_

* Conditional (or _if-then_) statements allow code to branch.
    * We can execute different chunks of code depending on which of two (or more) conditions is met.
* Why are they useful?
    * Epoching data.
    * Determining conditions.
    * Phenotyping cases.
    * Identifying Tx responders vs. non-responders.
    * Outlier detection.
    * etc etc etc.

```python
mango = 0.4

if mango < 0.50:
    print("mangoes are cheap these days")
```

```python
mango = 0.75

if mango < 0.5:
    print("mangoes are cheap these days")
elif mango < 1.0:
    print("get one mango from the store")
```

```python
mango = 1.50

if mango < 0.5:
    print("mangoes are cheap these days")
elif mango < 1.0:
    print("get one mango from the store")
else:
    print('meh, mangoes are gross anyways')
```

* What about when we have multiple conditions to meet?

```python
# trying to make a budget smoothie
mango = 1.50
blueberries = 0.80

# determining if they are both cheap enough to make my smoothie
if mango < 2 and blueberries < 1:
    print("its smoothie time")
else:
    print("i guess i will just drink water")
```

```python
# deciding whether to buy either fruit
mango = 1.50
blueberries = 0.49

if mango < 2 or blueberries < 1:
    print("we're getting a snack")
else:
    print("lets leave.")
```

* We can even combine _AND_ and _OR_

```python
budget = 3
mango_price = 2.50
banana_price = 1.50

mango_in_stock = False
banana_in_stock = True

if budget > mango_price and mango_in_stock or budget > banana_price and banana_in_stock:
    print("you can buy fruit")
else:
    print("better luck next time")
```

* Conditional presence in a list can be useful to...

## Nested control flows

### What is a 'control flow'?

* Iterations (_while, for_) and conditionals (_if-then, and-or_) are core computing constructs that allow us to control the _'flow'_ of our code.
    * i.e., the sequence of iterations, conditionals, etc. are common ways we tell the program the relative order to call functions or evaluate expressions.

### What is a 'nested control flow'?

* By the end of our programming journey, we'll likely have several control flows that need to be run, some in serial and some in parallel / nested fashion.
    * We can use iterations nested inside of conditionals, conditionals nested within iterations, conditionals nested within conditionals nested within iterations, etc. etc. etc.
 
* Example: Deciding about pricing after checking ripeness.

```python
# nested conditionals
mango_ripe = True
mango_price = 1.25

if mango_ripe:
    print("the mangoes look great today")
    if mango_price < 1:
        print("buy them all")
    elif mango_price < 2:
        print("just get one for dessert")
    else:
        print("too pricey...")
else:
    print("the mangoes are rock hard.")
    
    
```

* Example 2: Finding the best market given our price range  (nested conditional within a loop)

```python
# nested conditional within iteration

# Say we have observed a range of prices for mangoes at different supermarkets
markets = {
    "Smiths": [1.00, 2.50],
    "Whole_Foods": [2.00, 4.50],
    "Albertsons": [0.75, 2.00]
}

# reify default random number generator from np.random for price sampling
rng = np.random.default_rng()

# setting spending limit
spending_limit = 2.00

# Looping through stores
for store in markets:
    print(f"Checking {store}...")
    
    # Determining the current price
    min_price = min(markets[store])
    max_price = max(markets[store])
    cur_price = rng.uniform(min_price,max_price)

    # Conditional: Can we purchase given the current price?
    if cur_price < spending_limit:
        # print(f"Found a bargain: {cur_price}")
        # print(f"Found a bargain: {round(cur_price,2)}")
        # print("Found a bargain:",round(cur_price,2))
        print("Found a bargain: "+str(round(cur_price,2)))
    else:
        # print(f"Nothing cheap here: {cur_price}. Moving on.")
        # print(f"Nothing cheap here: {round(cur_price,2)}. Moving on.")
        # print("Nothing cheap here:",round(cur_price,2),". Moving on.")
        print("Nothing cheap here: "+str(round(cur_price,2))+". Moving on.")
```

### What if We Need to Repeat Some Process Many, Many Times?

* e.g. standard Plots, scoring questionnaires, transform data across different collections, computing stats, etc.
* To prevent this turning into wayyy too many lines of code: use _Functions_!

### Declaring Functions supports modular programming.
* Coding would be opaque & difficult to work with if we could only write sequences in serial from top to bottom (e.g., <a href="https://github.com/JeremyHogeveen/dspn_spring2026_github/blob/main/code/dspn_spring2026_week1_01222026.ipynb">our "imperative" PB&J instructions</a>).
* Functions are blocks of code that only run when explicitly called, which can be reconfigured in any number of ways (e.g., <a href="https://github.com/JeremyHogeveen/dspn_spring2026_github/blob/main/code/dspn_spring2026_week1_01222026.ipynb">our "declarative" PB&J instructions</a>).
    * Each function operates independently.
    * Functions then combined together in a workflow.
    * Allow you to build up a complex, but still (reasonably) well-organized codebase.

<img src="img/work_schematic.png" width="650">

### Coding Functions in Python

#### __def__

* Use __def__ and __:__ to create a user-defined function
    * Same indenting rules as iterations and conditionals
* __def__ doesn't actually _execute_ the function, it just creates it
    * Instead, subsequent to defining the function we execute it placing any associated inputs in parentheses -- __()__
    * Logic behind our code no longer needs to proceed in serial order.

### Example 1: Let's code a function that adds 5 to an input value

```python
# defining the function
def add_five(user_input):
    print(user_input + 5)
```

```python
add_five(6)
```
