# Section 1: The Jupyter "Notebook"

## 1.1 Working in a Jupyter "Kernel"

* When we launch _jupyter-lab_, we initiate a "kernel" for us to work in.
* Kernels are a fancy name for a way that we can interact with our computer using a specific programming language.
  * Jupyter or other IDEs (e.g. IPython) help us interface for an underlying code "interpreter" that takes the instructions we give it and spits out the appropriate data / figure object / etc.
  * Should be "Idle" when you are not running anything.
  * Do not run anything while the kernel is working (i.e., will say "Python (dspn_env) | "Busy").
  * If it seems frozen (not "Busy" but also not allowing you to type anything), try "Restart the kernel" option above.
  * When we are done working, after saving our work, select "File-->Shut Down" to close the kernel and free up your command prompt.
  * Kernels can be used for many different languages, most often it is used for Python, but in the future we will set up an R Kernel as well.

## 2.1 Writing Python code in Jupyter

* Python is an __object-oriented__ programming language
  * The things you create and/or read into Python are invariably some kind of __object__
* Python objects can be, ___variables___

* Variables are most often __integers, floats, or strings__

* Multiple variables can be assigned at once in Python

* Mathematical operators can be applied to numerical variables
  * Order of operations follows B() E** D/ M* A+ S-

* Division gives a few other options...
  * if you want the closest integer, and it's remainder you can do...

* Python objects might also be collections (e.g. data frames and lists)

* Python objects might also be plots
  * Won't get into this yet...

* With that in mind, lets write a code that can print "Hello World"

### Packages, sub-packages, and modules

* A ton of __packages__ have been developed
    * Some of them general purpose (e.g. Numpy, Pandas)
    * Some of them psych / neuro specific (e.g. psychopy, pygaze, MNE, Nipype, Brian, SPySort, etc etc etc)
* Those packages contain "sub-packages" which do a set of related things
* There are then individual "modules" or "functions" that do the stuff we want to do in Python
