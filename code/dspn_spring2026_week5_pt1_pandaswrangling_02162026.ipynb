# Section 2: Introducing _Numpy_

## 2.1 the _ndarray_
* The core data structure in Numpy is the n-dimensional array (or _ndarray_)
    * Function similarly to MATLAB arrays
    * An array with an arbitrary number of dimensions
    * Unlike _lists_, every element has to have the same data type

```python
# importing and name numpy package
import numpy as np

# storing a list of numbers
list_of_numbers = [1, 2, 3, 5, 6]
print(list_of_numbers)
# list_of_numbers.shape

# converting to a numpy array
array = np.array(list_of_numbers)
print(array)
array.shape
```

```python
# arrays can have multiple dimensions
list_of_numbers_1 = [4,3,9,1,6]
array_1 = np.array([list_of_numbers,list_of_numbers_1])
print(array_1)
```

```python
print("The shape of the first array is:",array.shape)
print("The shape of the second array is:",array_1.shape)
```

```python
# print("The dtype of the array is:",array.dtype)
print("The item size of the array is:",array.itemsize)
print("the strides of the first array is:",array.strides)
print("the strides of the second array is:",array_1.strides)

array_3 = np.array(['one','two','three','four','five'])
print(array_3)
array_3.itemsize

array_4 = np.array([1.11,2.22,3.33,4.44,5.55])
print(array_4)
array_4.itemsize

array_4[3]
```

## 3.1 Creating a data frame
* Can create a tidy data frame in Pandas using similar formatting to a _dictinary_ object.

```python
import pandas as pd

# create a pandas data frane
my_df = pd.DataFrame({'A': ['A0','A1','A2','A3']})
print(my_df)
```

```python
# can also add lists directly to the df
list_a = ['A0','A1','A2','A3']
list_b = ['B0','B1','B2','B3']
list_c = ['C0','C1','C2','C3']

my_df_2 = pd.DataFrame({'A': list_a,
                       'B': list_b,
                       'C': list_c})

print(my_df_2)
```

```python
# We can also create the data frame by assignign ...
# row values, ...
list_row1 = ['A0','B0','C0']
list_row2 = ['A1','B1','C1']
list_row3 = ['A2','B2','C2']
list_row4 = ['A3','B3','C3']
row_values = [list_row1, list_row2, list_row3, list_row4]
print(row_values)
# indices, ...
custom_indices = ['t1','t2','t3','t4']
print(custom_indices)
#  and columns separately.
col_headers = ['v1','v2','v3']
print(col_headers)

my_df_3 = pd.DataFrame(row_values,index=custom_indices,columns=col_headers)
print(my_df_3)
```

## 3.2 Reading an existing data frame

* More often, we have data we need to read in to wrangle.
    * Data can come from a local connected drive, a networked drive, or a remote server accessed over the web.
* Pandas specializes in reading in any data frame as long as it is _tab separated_ (aka "tabulated" or "tab-delimited")
    * comma separated values (csv)
    * tab separated values (tsv)
    * tab delimited text (txt)
    * Excel (.xlsx) can work, but requires an additional excel reader...
        * _conda install openpyxl_

```python
# reading csv from a web source
df_subinfo = pd.read_csv("https://yeatmanlab.github.io/AFQBrowser-demo/data/subjects.csv")
print(df_subinfo.head())
# print(df_subinfo.tail())
# print(df_subinfo.shape)
# print(df_subinfo.shape[0])
# print(df_subinfo.shape[1])
```

## 3.3 Reshaping data

### Melt (i.e. go from wide to long)

<img src="img/melt.png" width="400">

```python
# example:L pulling IQ subscales and makling long
df_subinfo_long = pd.melt(df_subinfo,
                          id_vars=['subjectID','Age','Gender','Handedness','IQ'],
                         value_name = 'IQ_score',
                         var_name = 'IQ_scale')
print(df_subinfo_long)
```

### Pivot (i.e. go from long to wide)

<img src="img/pivot.png" width="600">

```python
# example: moving IQ subcale scores back to wide format
df_subinfo_wide = df_subinfo_long.pivot(index = ['subjectID','Age','Gender','Handedness','IQ'],
                                       columns = 'IQ_scale',
                                       values='IQ_score')
# print(df_subinfo)
print(df_subinfo_wide)
print(df_subinfo_wide.reset_index().rename_axis(None,axis=1))
```
