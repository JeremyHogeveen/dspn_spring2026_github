### Task details
* 2-alternative forced choice (Same or Different)
* Stimulus Types: Same image or mirror image
* Rotation Conditions: 0, 80, 120, or 160 degree rotation
* In the dataset, accuracy for each condition is represented by the variables: `MenRot_0_CORRpc`, `MenRot_80_CORRpc`, `MenRot_120_CORRpc`, and `MenRot_160_CORRpc`.

## Assignment #5 : __Mixed Model Fun__!

Load in the dataset (`data/data_mod_exercise/GSP_cleaned_long.csv`) and use the techniques we learned over the last few weeks to compare a traditional ANOVA with a Linear Mixed Model.

### Core Objectives: Model Face-Off
1. **The Classic Mixed ANOVA:** Use the `aov_ez()` function from the `afex` package to run a mixed ANOVA. Your within-subject factor is `Angle`, and your between-subjects factor is `Group`.
2. **The LMM Equivalent:** Fit a Linear Mixed Model (using `lmer()`) testing the exact same main effects and interaction. Include a **random intercept** for each Subject. 
3. **Compare & Contrast:** Generate the omnibus tables for both models. Do they yield similar F-values and p-values for the main effects and interaction? Use `emmeans` to extract the pairwise comparisons for the `Angle` effect from one of the models. 

### Bonus Objectives: Why we use LMMs
* **Bonus 1 (Random Slopes):** One major advantage of `lmer` over standard ANOVA is the ability to let within-subject effects vary by individual. Update your `lmer` model to include a *random slope* of `Angle` for each subject (`Angle | Subject_ID`). Does the model successfully converge?
* **Bonus 2 (Model Diagnostics):** Check the assumption of normality for your `lmer` model's residuals. Do the residuals look normally distributed, or are there extreme values/skewness?
* **Bonus 3 (Robust Modeling):** Another massive advantage of the LMM framework is robustness. Refit your model using `rlmer()` from the `robustlmm` package to actively down-weight outliers. Generate the omnibus Wald-test table for this robust model. Does down-weighting the outliers change your inferences compared to your original `aov_ez` results?
